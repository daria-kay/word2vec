## Training word2vec embeddings

```python
from embedding import Word2VecDatasetBuilder, Word2Vec
```

Preparing 3 different datasets from a specified corpus: each dataset contains word pairs generated using a different context window size. The amount of negative samples is set to 20 by default.

```python
RAW_DATA = "data/corpora.txt"
dataset_builder = Word2VecDatasetBuilder()

default_directory = 'data/default_window'
dataset_builder.build_dataset(RAW_DATA, default_directory, window_size=5)

small_directory = 'data/small_window'
dataset_builder.build_dataset(RAW_DATA, small_directory, window_size=3)

big_directory = 'data/big_window'
dataset_builder.build_dataset(RAW_DATA, big_directory, window_size=7)
```

### Training models on each dataset

Simple SGD is used during model training. Since all matrix operations use numpy arrays the elapsed time may be significant

```python
model = Word2Vec.train(default_directory, embedding_size=100, epoches=10, lr=1e-2)
model.save_pretrained(default_directory)
```

```python
model = Word2Vec.train(small_directory, embedding_size=100, epoches=5, lr=1e-2)
model.save_pretrained(small_directory)
```

```python
model = Word2Vec.train(big_directory, embedding_size=100, epoches=5, lr=1e-2)
model.save_pretrained(big_directory)
```
